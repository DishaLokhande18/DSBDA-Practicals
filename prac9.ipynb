```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data= sns.load_dataset('titanic')
```

```python
sns.boxplot(x='sex',y='age',hue='survived',data=data)
```

```python
plt.title("Age Distribution by Gender and Survival")
plt.show()
```

```python
plt.title("Age Distribution by Gender and Survival")
sns.boxplot(x='sex',y='age',hue='survived',data=data)
plt.show()
```
