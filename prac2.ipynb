```python
import numpy as np
import pandas as pd
```

```python
df=pd.DataFrame({
    'name':['a','b','c','d','e'],
    'math':[98,60,np.nan,200,79],
    'science':[np.nan,78,98,87,90]
})
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
df['math'].fillna(df['math'].mean(),inplace=True)
```

```python
df['science'].fillna(df['science'].mean(),inplace=True)
```

```python
q1=df['math'].quantile(0.25)
q3=df['math'].quantile(0.75)
iqr=q3-q1
lower=q1-1.5*iqr
upper=q3+1.5*iqr
print("lower limit:", lower)
print("upper limit:", upper)
```

```python
df['math'] = np.where(df['math'] < lower, lower, df['math'])
df['math'] = np.where(df['math'] > upper, upper, df['math'])
```

```python
df['Maths_norm'] = (df['math'] - df['math'].min()) / (df['math'].max() - df['math'].min())
```
